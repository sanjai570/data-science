## re patterns

```python
## \d --extract only numbers

re.findall('\d+',str1)
```

```python
## \D --extract except  number

re.findall('\D+',str1)
```

```python
## \s -- exrtact space alone
re.findall('\s',str1)
```

```python
## \S -- exrtact except space 
re.findall('\S+',str1)
```

```python
## \w --extract except spaces and non-alphanumeric
re.findall('\w+',str1)
```

```python
## \W --extract except spaces and non-alphanumeric
re.findall('\W+',str1)
```
