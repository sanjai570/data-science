```python
import pandas as pd
import numpy as np
import warnings
warnings.filterwarnings('ignore')
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('sample_data2.csv')
```

```python
df.drop_duplicates(inplace = True,ignore_index = True)
```

```python
df
```

```python
## handlings missing values
df.isnull().sum().sum()/df.size*100
```

```python
df.dropna() ## to remove missing values
```

```python
df.Salary.mean()
```

```python
df.Qualification.mode()
```

```python
df.fillna({'Salary': df.Salary.mean(),'Qualification':df.Qualification.mode()[0]},inplace = True)
```

```python
df.isnull().sum().sum()/df.size*100
```

```python
from sklearn.preprocessing import StandardScaler,MinMaxScaler,RobustScaler
```

```python
std_sca = StandardScaler()
min_max = MinMaxScaler()
rob_sca = RobustScaler()
```

```python
std_sca.fit_transform(df[['Salary']])
```

```python
min_max.fit_transform(df[['Salary']])
```

```python
rob_sca.fit_transform(df[['Salary']])
```

```python
df[['Salary']] = std_sca.fit_transform(df[['Salary']])
```

```python
df
```

```python
from sklearn.preprocessing import OrdinalEncoder , LabelEncoder,OneHotEncoder
```

```python
ord_enc = OrdinalEncoder(categories = [['Bachelors','Masters']])
df[['Qualification']] = ord_enc.fit_transform(df[['Qualification']])
```

```python
lab_enc = LabelEncoder()
lab_enc.fit_transform(df[['Qualification']])
```

```python
df
```

```python
pd.get_dummies(df, columns=['Title'],dtype = 'int') 
```
