```python
import matplotlib.pyplot as plt
import seaborn as sns
import warnings 
warnings.filterwarnings('ignore')
```

```python
Exp=[1,2,4,6,10,15]
Sal=[1000,2500,3000,5000,10000,15000]
```

```python
## scatter plot
plt.scatter(Exp,Sal)
```

```python
plt.scatter(Exp,Sal,marker = '*')
```

```python
plt.figure(figsize = (5,5))
plt.title('exp X sal' , color = 'green' , size = 14)
plt.scatter(Exp,Sal,color = 'red')
plt.xlabel('experince',color = 'green',size = 12)
plt.ylabel('salary',color = 'green',size = 12)
plt.grid()
```

```python
plt.figure(figsize = (5,5))
plt.title('exp X sal' , color = 'green' , size = 14)
plt.plot(Exp,Sal,color = 'red',marker = 'o',linestyle='--',markerfacecolor='green',lw = 2)
plt.xlabel('experince',color = 'green',size = 12)
plt.ylabel('salary',color = 'green',size = 12)
plt.grid()
```

```python
#scatter plot

sns.scatterplot(data = df,x = 'carat' , y = 'price' ,color = 'red')
```

```python
df.cut.values
```

```python
df.cut.value_counts()
```

```python
cntplt =  sns.countplot(data = df, x = 'cut')
cntplt.bar_label(cntplt.containers[0])
```

```python
# histogram 
sns.histplot(data = df, x = 'carat' ,kde = True, color = 'red',bins = 100 ) #kde kernel distributed estimate 
```

```python
## density plot
sns.kdeplot(data = df, x = 'carat')
```

```python
sns.kdeplot(data = df, x = 'carat',fill = True,color = 'grey',alpha = 0.4)
```

```python
# to seperate numerical and catogorical columns

num_col = df.select_dtypes(include = ['int','float']).columns
print(num_col)

cat_col = df.select_dtypes(include = 'object').columns
print(cat_col)
```

```python
## supplot -- histogram

for i in range(len(num_col)):
    print(i)
```

```python
plt.figure(figsize = (15,15))
for i in range(len(num_col)):
    plt.subplot(4,2,i+1)
    sns.histplot(x = df[num_col[i]])
    plt.tight_layout()
```

```python
plt.figure(figsize = (12,7))
for i in range(len(num_col)):
    plt.subplot(4,2,i+1)
    sns.kdeplot(x = df[num_col[i]])
    plt.tight_layout()
```

```python
plt.figure(figsize = (12,7))
for i in range(len(num_col)):
    plt.subplot(4,2,i+1)
    sns.boxplot(x = df[num_col[i]])
    plt.tight_layout()
```

```python
plt.figure(figsize = (12,7))
for i in range(len(cat_col)):
    plt.subplot(4,2,i+1)
    sns.countplot(x = df[cat_col[i]])
    plt.tight_layout()
```
