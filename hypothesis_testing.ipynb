```python
import numpy as np
import scipy.stats as stats
import statsmodels
import warnings 
warnings.filterwarnings('ignore')
```

```python
## given data 
pop_mean = 168
pop_std_dev = 3.9
n = 36
x_bar = 169.5
## ci - 95% , alpha - 5%
```

```python
## H0 - pop_mean = 168cm
## Ha - pop_mean != 168cm
zscore = (x_bar - pop_mean)/(pop_std_dev/np.sqrt(n))
zscore
```

```python
#### From z-table, z-critical=1.96
#### Since z_critical < z_score , reject H0
#### Pop_mean !=168cm
```

```python
stats.norm.ppf(0.975)
```

```python
s = pop_std_dev/np.sqrt(n)
s
```

```python
sample_data = stats.norm.rvs(size = n,loc = x_bar, scale = s)
sample_data
```

```python
from statsmodels.stats.weightstats import ztest
zscore,pvalue = ztest(x1 = sample_data,value = pop_mean)
print(zscore ,pvalue)
```
